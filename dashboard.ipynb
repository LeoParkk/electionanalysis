```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns


df = pd.read_csv("HoC-GE2024-results-by-constituency.csv")
major_parties = ['Lab', 'Con', 'LD', 'RUK', 'Green', 'SNP']
df.info()
```

```python
# Question 1
```

```python
# Question 2 - efficency (votes, to seats won)
votes = df[major_parties].sum() / df['Valid votes'].sum() * 100
seats = df['First party'].value_counts()[major_parties] / len(df) * 100

pd.DataFrame({'Votes %': votes, 'Seats %': seats}).plot.bar(rot=0)
plt.title('Vote share vs seat share')
plt.show()
```
